# Count the number of years in which 10-k reports have been filed within the past 5 years





```python
import os
import csv

# Specify the parent folder path
parent_folder = "/content/drive/MyDrive/Advanced NLP/Assignments/Assignment_1/data files/sec-edgar-filings"

# Specify the CSV file name
csv_file = "/content/cik_years_list.csv"

# List to store the extracted data
data = []

# Iterate over immediate subfolders (CIK folders)
for cik_folder in os.listdir(parent_folder):
    cik_path = os.path.join(parent_folder, cik_folder)

    # Ensure it's a directory
    if os.path.isdir(cik_path):
        ten_k_path = os.path.join(cik_path, "10-K")

        # Check if the "10-K" folder exists
        if os.path.isdir(ten_k_path):
            # Get the subfolder names under "10-K"
            years = [folder for folder in os.listdir(ten_k_path) if os.path.isdir(os.path.join(ten_k_path, folder))]
            years_str = ",".join(years)  # Convert list to comma-separated string

            # Append data
            data.append([cik_folder, years_str])

# Save the extracted data to a CSV file
with open(csv_file, mode="w", newline="") as file:
    writer = csv.writer(file)
    writer.writerow(["CIK", "Years"])  # Write header
    writer.writerows(data)  # Write data

print(f"Extracted data has been saved to '{csv_file}'")
```

```python
import pandas as pd
import re

# Load the CSV file
input_csv = "/content/cik_years_list.csv"  # Replace with your actual file name
output_csv = "/content/cik_extracted_years.csv"

# Read the CSV file
df = pd.read_csv(input_csv)

# Function to extract unique years from the "Years" column
def extract_years(years_str):
    years = re.findall(r'-(\d{2})-', years_str)  # Extract two-digit years
    return ",".join(sorted(set(years)))  # Remove duplicates and sort

# Apply the function to extract years
df["Extracted Years"] = df["Years"].astype(str).apply(extract_years)

# Select only the CIK and Extracted Years columns
df = df[["CIK", "Extracted Years"]]

# Save to a new CSV file
df.to_csv(output_csv, index=False)

print(f"Extracted years have been saved to '{output_csv}'")
```

```python
import pandas as pd

# Load the CSV file
input_csv = "/content/cik_extracted_years.csv"  # Replace with your actual file name
output_csv = "/content/cik_10k_filed_years.csv"

# Read the CSV file
df = pd.read_csv(input_csv)

# Define the years to check
years = ["20", "21", "22", "23", "24"]  # Corresponds to 2020-2024

# Function to check if a year is filed
def check_filed(years_str, year):
    return "Filed" if year in years_str.split(",") else "Not Filed"

# Create new columns for each year
for year in years:
    df[f"20{year}"] = df["Extracted Years"].apply(lambda x: check_filed(x, year))

# Drop the "Extracted Years" column
df.drop(columns=["Extracted Years"], inplace=True)

# Save to a new CSV file
df.to_csv(output_csv, index=False)

print(f"Transformed data has been saved to '{output_csv}'")
```
