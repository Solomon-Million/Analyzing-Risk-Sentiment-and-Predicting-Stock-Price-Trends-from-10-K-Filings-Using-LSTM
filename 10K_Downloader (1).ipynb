```python
from sec_edgar_downloader import Downloader
from tqdm import tqdm
import pandas as pd
import os

# Mount Google Drive
from google.colab import drive
drive.mount('/content/drive')

# Path to CSV file in Google Drive
csv_file = "/content/drive/MyDrive/Advanced NLP/Assignments/Assignment_1/CSV_Files/CIK_unique_pairs.csv"  # Update with actual CSV file path

# Load CIK list from CSV
df = pd.read_csv(csv_file)

# Ensure 'cik' column exists and is formatted correctly
df['cik'] = df['cik'].astype(str).str.zfill(10)  # Ensuring CIKs are 10 digits

# Convert to list
cik_list = df['cik'].tolist()

# Set download directory
download_dir = "/content/drive/MyDrive/Advanced NLP/Assignments/Assignment_1/10K_files_2000_2020"
os.makedirs(download_dir, exist_ok=True)

# Initialize SEC EDGAR Downloader
dl = Downloader(download_folder=download_dir, email_address="million.solomon@torontomu.ca", company_name="Toronto Metropolitan University")

# Define filing types to download
filing_types = ["10-K"]

# Function to download filings
def download_filings(cik_list, filing_types):
    for cik in tqdm(cik_list, desc="Processing CIKs"):
        for filing_type in tqdm(filing_types, desc=f"Downloading filings for CIK {cik}", leave=False):
            try:
                print(f"Downloading {filing_type} for CIK {cik}...")
                dl.get(filing_type, cik, after="2000-01-01", before="2020-02-25")
            except Exception as e:
                print(f"Failed to download {filing_type} for CIK {cik}: {e}")
    print("Download completed!")

# Start the download process
print("Starting the download process...")
download_filings(cik_list, filing_types)
```
